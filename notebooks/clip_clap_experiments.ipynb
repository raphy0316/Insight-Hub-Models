# CLIP × CLAP Experiments

Minimal demo that retrieves music given image keywords or images. This notebook uses text fallback if CLIP/CLAP are unavailable.

```python
from src.model.clip_clap import ClipClapModel

# Load model from config
model = ClipClapModel.from_config("config/default.yaml")

# Sample music items
music_items = [
    {
        "id": "track_001",
        "title": "Midnight Drive",
        "artists": ["Luna Wave"],
        "album": "Nocturne",
        "genres": ["synthwave", "electronic"],
        "editorial_tags": ["retro", "night city", "neon"],
        "lyrics": "Riding through the empty roads under neon lights...",
        "summary": "Retro electronic vibe suitable for night-time cityscapes.",
        # "audio_path": "path/to/audio.wav",  # optional for CLAP
    },
    {
        "id": "track_002",
        "title": "Pastel Morning",
        "artists": ["Sunny Garden"],
        "album": "Daybreak",
        "genres": ["lofi", "chill"],
        "editorial_tags": ["calm", "soft", "morning"],
        "lyrics": None,
        "summary": "Soft lofi beats for calm mornings.",
    },
]

# Build index
model.prepare_music_index(music_items)

# Query by image keywords (text pivot)
image_keywords = ["neon", "night city", "retro", "car driving"]
results = model.search_by_image_keywords(image_keywords, top_k=5)
results[:3]

# Note: To search with image files directly, use:
# results = model.search_by_image_paths(["path/to/image1.jpg"], top_k=5)
```
